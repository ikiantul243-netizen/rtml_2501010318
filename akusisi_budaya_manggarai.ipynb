```python

import requests
import json

url = "https://overpass-api.de/api/interpreter"

query = """
[out:json][timeout:60];
area["ISO3166-2"="ID-NT"]->.ntt;
(
  nwr["historic"="house"](area.ntt);
  nwr["historic"="building"](area.ntt);
  nwr["building:architecture"](area.ntt);
);
out center tags;
"""

headers = {
    "User-Agent": "TugasAkuisisiDataINSTIKI/1.0 (email-kampus-anda)"
}

try:
    response = requests.post(
        url,
        data={"data": query},
        headers=headers,
        timeout=90
    )

    response.raise_for_status()

    hasil = response.json()

    print("Status:", response.status_code)
    print("Jumlah objek ditemukan:", len(hasil["elements"]))

except requests.exceptions.RequestException as e:
    print("Terjadi kesalahan:", e)
```
